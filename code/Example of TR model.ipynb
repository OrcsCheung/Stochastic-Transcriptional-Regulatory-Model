```python
import pandas as pd

gene = pd.read_csv('/content/top_data.csv')
```

```python
sample1=gene[["CTCTACGTCAGATAAG-1"]].astype(float)
```

```python
sample1=sample1.to_numpy()
sample1=np.float64(sample1)
```

```python
sample=[]
for i in range(len(sample1)):
    a=sample1[i].item()
    sample.append(a)
```

```python
iteration=len(sample)
```

Step 2: Set the value of 5 parameters, time step and changes of reactions.

```python
#test5
step= 500
Kon = 0.9
Koff = 10
psi = 5000
degradation_rates = np.random.lognormal(mean=-0.45, sigma=1.38, size=step)
transcription_rates = np.random.lognormal(mean=0.27, sigma=1.95, size=step)
s=transcription_rates*(Kon+Koff)/(Kon*psi)
intial_I = 0.01
c1=np.random.lognormal(mean=-7.51, sigma=1.78, size=step)
c2=np.random.lognormal(mean=-6.20, sigma=2.32, size=step)
d=np.random.lognormal(mean=6.11, sigma=4.44, size=step)
a=np.random.lognormal(mean=0.06, sigma=1.37, size=step)
```

```python
gene1=[]
gene2=[]
```

Step 3: import only 1 sample to the model and run the model twice, and then save the results as gene1 and gene2.

```python
for j in range(iteration):



    A = 0
    I = intial_I
    X = sample[j]

    t_values = [0]
    A_values = [A]
    I_values = [I]
    X_values = [X]


    for i in range(step):

        R = np.array([Koff * A_values[i], Kon * I_values[i],psi * s[i] * A_values[i], degradation_rates[i] * X_values[i]])
        total_rate = np.sum(R)

        tau = 1/total_rate * np.log(1/np.random.random())
        t_values.append(t_values[-1] + tau)

        reaction = np.random.choice([0, 1, 2, 3], p=R / total_rate)

        if reaction == 0:
            A -= d[i]
            I += d[i]
        elif reaction == 1:
            A += a[i]
            I -= a[i]
        elif reaction == 2:
            X += c1[i]
        elif reaction == 3:
            X -= c2[i]

        A_values.append(A)
        I_values.append(I)
        X_values.append(X)

        if X_values[-1]< 0:
            X_values[-1]=0


        if A_values[-1]< 0:
            A_values[-1]=0

        if I_values[-1]< 0:
            I_values[-1]=0




    gene1.append([X_values[-1]])
```
